### 🤔 Por que o Python puro não é suficiente?

Imagine que você quer somar `10` a cada uma das temperaturas registradas por um sensor, guardadas numa lista:


````python
temperaturas = [78.2, 79.0, 77.5, 80.1, 78.9]

# Isso NÃO funciona como muita gente espera:
# temperaturas + 10   -> geraria erro!

# Precisamos de um laço for para fazer isso "na mão":
temperaturas_ajustadas = []
for temp in temperaturas:
    temperaturas_ajustadas.append(temp + 10)

print(temperaturas_ajustadas)
````

**Perceba que** precisamos escrever várias linhas de código só para somar um número a cada item da lista. Com listas muito grandes (milhares ou milhões de leituras), esse processo além de trabalhoso, é **lento**, porque o Python percorre item por item, um de cada vez.

O **NumPy** (Numerical Python) é uma biblioteca criada exatamente para resolver esse problema: ela permite fazer operações matemáticas em **todos os elementos de uma vez**, de forma muito mais rápida e com muito menos código. É amplamente usada na indústria em qualquer lugar onde existam grandes volumes de dados numéricos: sensores, controle de processos, visão computacional e, claro, como base para outras bibliotecas de IA (como Pandas, Scikit-learn e TensorFlow).

### 🏠 Analogia do cotidiano

Fazer contas com listas comuns é como somar um número a cada peça de uma bandeja de peças, uma de cada vez, pegando-a com a mão. Usar o NumPy é como ter uma máquina que faz a mesma operação em todas as peças da bandeja **simultaneamente**, num único movimento.

## 1. Importando o NumPy e criando um Array

### 📘 Conceito teórico

Um **array** do NumPy é parecido com uma lista, mas otimizado para cálculos matemáticos rápidos com números.

Para usar o NumPy, primeiro precisamos **importar** a biblioteca — um comando que "avisa" ao Python que queremos usar ferramentas que não vêm por padrão na linguagem.

### 🔎 Exemplo extremamente simples


````python
import numpy as np

temperaturas = np.array([78.2, 79.0, 77.5, 80.1, 78.9])
print(temperaturas)
print(type(temperaturas))
````

### 🔍 Explicação linha por linha

| Código | O que significa |
|---|---|
| `import numpy as np` | Importa a biblioteca NumPy e a "apelida" de `np`, para não precisarmos digitar `numpy` inteiro toda vez. Esse apelido é uma convenção usada por praticamente todo mundo que programa em Python. |
| `np.array([...])` | Transforma uma lista comum num **array do NumPy**. |
| `type(temperaturas)` | Confirma que o tipo agora é `numpy.ndarray`, diferente de uma lista comum. |

**Neste momento não se preocupe em decorar** todos os detalhes do NumPy — o importante é entender que arrays permitem operações matemáticas muito mais diretas.

### 🏭 Exemplo aplicado à indústria: agora a soma funciona direto


````python
temperaturas_ajustadas = temperaturas + 10
print(temperaturas_ajustadas)
````

**Observe que**, com o array do NumPy, conseguimos somar `10` a **todos os elementos de uma vez**, sem precisar de um `for`. Essa capacidade é chamada de **operação vetorizada**.

### 📦 O que você aprendeu até aqui?

| Conceito | Em palavras simples |
|---|---|
| NumPy | Biblioteca para cálculos numéricos rápidos e eficientes |
| Array | Estrutura parecida com lista, mas otimizada para matemática |
| Operação vetorizada | Aplicar uma operação a todos os elementos de uma vez, sem `for` |

## 2. Operações com Arrays

### 📘 Conceito teórico

Além de somar um número, podemos fazer operações entre **dois arrays inteiros**, elemento por elemento, e também aplicar funções matemáticas prontas do NumPy.

### 🏠 Analogia do cotidiano

É como comparar duas fileiras de caixas numa esteira, uma com o que foi **planejado** produzir e outra com o que foi **realmente** produzido: você compara a primeira caixa com a primeira, a segunda com a segunda, e assim por diante — nunca a terceira caixa planejada com a quinta caixa real.

### 🔎 Exemplo extremamente simples


````python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print(a + b)
print(a * b)
````

**Perceba que** a soma `a + b` soma o primeiro elemento de `a` com o primeiro de `b`, o segundo com o segundo, e assim por diante — não é uma soma "de tudo com tudo".

### 🏭 Exemplo aplicado à indústria


````python
producao_planejada = np.array([500, 520, 480, 510])
producao_real = np.array([470, 515, 460, 500])

diferenca = producao_real - producao_planejada
print("Diferença por dia:", diferenca)

percentual_atingido = (producao_real / producao_planejada) * 100
print("Percentual da meta atingido:", percentual_atingido)
````

### 📦 O que você aprendeu até aqui?

| Conceito | Em palavras simples |
|---|---|
| Operação entre arrays | Combina elemento por elemento, na mesma posição |

## 3. Estatística Básica com NumPy

### 📘 Conceito teórico

O NumPy já vem com funções prontas para calcular estatísticas básicas: média, valor máximo, valor mínimo e desvio padrão (uma medida de quanto os valores variam em torno da média).

### 🏠 Analogia do cotidiano

É como um médico resumindo os sinais vitais de um paciente em poucos números (pressão média, temperatura, batimentos) em vez de mostrar todas as centenas de medições do dia. As estatísticas resumem rapidamente "como está indo" um conjunto grande de dados.

### 🔎 Exemplo extremamente simples


````python
notas = np.array([7, 8, 6, 9, 10])

print("Média:", np.mean(notas))
print("Máximo:", np.max(notas))
print("Mínimo:", np.min(notas))
````

### 🏭 Exemplo aplicado à indústria


````python
vibracao = np.array([2.1, 2.3, 2.0, 5.8, 2.2, 2.1])

media_vibracao = np.mean(vibracao)
desvio_padrao = np.std(vibracao)

print("Média de vibração:", media_vibracao)
print("Desvio padrão:", desvio_padrao)
````

**Perceba que** o valor `5.8` está bem distante dos demais — isso é o que chamamos de um **outlier** (valor fora do padrão), e pode indicar, por exemplo, um problema pontual na máquina naquele instante. Vamos explorar outliers com mais profundidade na Missão 9 (EDA).

### 📦 O que você aprendeu até aqui?

| Função | O que calcula |
|---|---|
| `np.mean()` | Média |
| `np.max()` / `np.min()` | Valor máximo / mínimo |
| `np.std()` | Desvio padrão (variação em torno da média) |

## 4. Gerando Dados Simulados

### 📘 Conceito teórico

O NumPy também permite **gerar dados aleatórios**, muito útil para simular sensores quando ainda não temos dados reais disponíveis (como faremos ao longo deste curso, usando a fábrica fictícia Alpha-SP Indústria).

### 🏠 Analogia do cotidiano

É como testar o painel de controle de uma máquina nova com um "boneco de teste" antes de ela ser ligada de verdade na linha de produção — os dados simulados fazem esse papel de "teste antes do sensor real chegar".

### 🔎 Exemplo extremamente simples


````python
np.random.seed(42)
valores_aleatorios = np.random.randint(1, 10, 5)
print(valores_aleatorios)
````

### 🔍 Explicação linha por linha

| Código | O que significa |
|---|---|
| `np.random.seed(42)` | Define uma "semente" fixa para os números aleatórios. Isso garante que, toda vez que você executar o código, os **mesmos** números "aleatórios" apareçam — muito útil para exercícios e para reproduzir resultados. |
| `np.random.randint(1, 10, 5)` | Gera 5 números inteiros aleatórios entre 1 (incluso) e 10 (exclusivo). |

### 🏭 Exemplo aplicado à indústria: simulando leituras de um sensor


````python
np.random.seed(10)
leituras_sensor = np.random.normal(loc=75, scale=2, size=10)
print(np.round(leituras_sensor, 1))
````

**Observe que** usamos `np.random.normal()` para simular leituras que seguem uma distribuição parecida com a de sensores reais: a maioria dos valores fica próxima da média (`loc=75`), com uma variação típica (`scale=2`). Usamos `np.round(..., 1)` para arredondar os valores para uma casa decimal, facilitando a leitura.

### 📦 O que você aprendeu até aqui?

| Conceito | Em palavras simples |
|---|---|
| `np.random.seed()` | Garante que números "aleatórios" se repitam da mesma forma em execuções diferentes |
| `np.random.randint()` | Gera números inteiros aleatórios num intervalo |
| `np.random.normal()` | Gera números aleatórios seguindo uma distribuição em torno de uma média |

## 🧪 Exercício guiado

Vamos simular um turno de leituras de temperatura e calcular estatísticas básicas:


````python
np.random.seed(1)
temperaturas_turno = np.random.normal(loc=80, scale=3, size=20)

print("Média do turno:", np.round(np.mean(temperaturas_turno), 2))
print("Temperatura máxima:", np.round(np.max(temperaturas_turno), 2))
print("Temperatura mínima:", np.round(np.min(temperaturas_turno), 2))
````

1. **Esquecer de importar o NumPy antes de usá-lo.**

2. **Tentar somar arrays de tamanhos diferentes.**
   Somar um array de 3 elementos com outro de 5 elementos gera erro (na maioria dos casos) — os arrays precisam ser compatíveis em tamanho.

3. **Confundir array do NumPy com lista comum.**
   Embora pareçam parecidos ao serem exibidos, uma lista comum não permite operações matemáticas diretas como `lista + 10`.

4. **Esquecer o `np.random.seed()` e depois se surpreender com resultados diferentes a cada execução.**
   Sem a semente fixa, os números aleatórios mudam toda vez que o código roda — o que pode ser desejado ou não, dependendo do objetivo.
