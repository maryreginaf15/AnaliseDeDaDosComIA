```python
# Faça um programa que leia o peso de 12 pessoas.
# No final, mostre qual foi o maior e o menor peso lidos.

maior = float('-inf')
menor = float('inf')

for i in range(1,13):
    peso = float(input(f"digire o {i} /12 peso: "))

    if peso > maior:
        maior = peso
        if peso < menor:
            menor = peso


    print(f"Maior peso: {maior} kg")
    print(f"Menor peso: {menor} kg")
```

```python
#solicitar 10 numeros inteiros para o usuario
#adicionar os numeros negativos em uma lista, os pares positivos em outra
# e os impares positivos em outra


listaNegativa = []
listaPar = []
listaImpar = []

for i in range(1,11):
    num = int(input(f"digite um numero inteiro positivo ou negativo: {i}/10: "))

if num < --0:
        listaNegativa.append(num)
elif num % 2 == 0:
        listaPar.append(num)
else:
        listaImpar.append(num)

if len(listaNegativa) == 0:
    listaNegativa = "nao foram digitados numeros negativos"
if len(listaPar) == 0:
    listaPar = "nao foram digitados numeros pares"
if len(listaImpar) == 0:
    listaImpar = "nao foram digitados numeros impares"

print("numeros negativos: ", listaNegativa)
print("numeros pares : ", listaPar)
print("numeros negativos: ", listaImpar)

```
