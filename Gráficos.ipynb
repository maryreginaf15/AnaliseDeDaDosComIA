```python
import pandas as pd
import plotly.express as px

receitas_lojas = pd.DataFrame({
    'nome_loja': ['Loja A', 'Loja B', 'Loja C', 'Loja D'],
    'valor_final': [15000, 22000, 18000, 30000]
})

fig = px.bar(
    receitas_lojas,
    x='nome_loja',
    y='valor_final',
    title='Receitas por Loja',
    color='valor_final',
    text='valor_final',
    color_continuous_scale=px.colors.diverging.RdBu_r
)

fig.update_traces(
    textposition='outside',
    texttemplate='R$ %{text:,.0f}'
)

fig.show()
```

```python

import plotly.express as px
import plotly.graph_objects as go

# Criar uma cópia e ordenar as lojas pela receita
df = receitas_lojas.sort_values(
    by='valor_final',
    ascending=False
).copy()

# Calcular participação percentual de cada loja
df['participacao'] = (
    df['valor_final'] / df['valor_final'].sum()
) * 100

# Calcular receita média
media = df['valor_final'].mean()

# Criar o gráfico
fig = px.bar(
    df,
    x='nome_loja',
    y='valor_final',
    title='Receitas por Loja',
    color='valor_final',
    text='valor_final',
    custom_data=['participacao'],
    color_continuous_scale='Blues',
    labels={
        'nome_loja': 'Loja',
        'valor_final': 'Receita (R$)',
        'participacao': 'Participação'
    }
)

# Personalizar os valores nas barras e o tooltip
fig.update_traces(
    texttemplate='R$ %{y:,.2f}',
    textposition='outside',
    hovertemplate=(
        '<b>%{x}</b><br>'
        'Receita: R$ %{y:,.2f}<br>'
        'Participação: %{customdata[0]:.2f}%'
        '<extra></extra>'
    ),
    cliponaxis=False
)

# Adicionar linha da receita média
fig.add_hline(
    y=media,
    line_dash='dash',
    line_color='red',
    annotation_text=f'Média: R$ {media:,.2f}',
    annotation_position='top left'
)

# Ajustar layout
fig.update_layout(
    template='plotly_white',
    xaxis_title='Nome da loja',
    yaxis_title='Receita final (R$)',
    coloraxis_colorbar_title='Receita (R$)',
    uniformtext_minsize=9,
    uniformtext_mode='hide'
)

fig.show()
```

```python

import pandas as pd
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# ==========================================
# 1. CRIAR DADOS FICTÍCIOS
# ==========================================

df_final = pd.DataFrame({
    'dt_venda': [
        '2025-01-05', '2025-01-18',
        '2025-02-08', '2025-02-20',
        '2025-03-10', '2025-03-22',
        '2025-04-03', '2025-04-19',
        '2025-05-07', '2025-05-25',
        '2025-06-11', '2025-06-27',
        '2025-07-04', '2025-07-21',
        '2025-08-09', '2025-08-23',
        '2025-09-06', '2025-09-18',
        '2025-10-02', '2025-10-24',
        '2025-11-08', '2025-11-20',
        '2025-12-05', '2025-12-19'
    ],
    'valor_final': [
        4500, 5200,
        4800, 6100,
        5500, 7200,
        6800, 7500,
        7200, 8100,
        8500, 9200,
        7800, 9500,
        9000, 10500,
        9800, 11200,
        11000, 12500,
        13500, 14800,
        16000, 18500
    ]
})

# Converter a coluna de datas
df_final['dt_venda'] = pd.to_datetime(
    df_final['dt_venda']
)

# ==========================================
# 2. AGRUPAR POR ANO E MÊS
# ==========================================

df_final['ano - mes'] = (
    df_final['dt_venda'].dt.to_period('M')
)

df_fat_mes = (
    df_final.groupby('ano - mes')['valor_final']
    .sum()
    .reset_index()
)

# Converter o período para data
df_fat_mes['ano - mes'] = (
    df_fat_mes['ano - mes'].dt.to_timestamp()
)

# ==========================================
# 3. CALCULAR INDICADORES
# ==========================================

# Faturamento do mês anterior
df_fat_mes['mes_anterior'] = (
    df_fat_mes['valor_final'].shift(1)
)

# Diferença em reais em relação ao mês anterior
df_fat_mes['diferenca_R$'] = (
    df_fat_mes['valor_final']
    - df_fat_mes['mes_anterior']
)

# Variação percentual mensal
df_fat_mes['variacao_%'] = (
    df_fat_mes['valor_final'].pct_change() * 100
)

# Média mensal do faturamento
media_mensal = df_fat_mes['valor_final'].mean()

# Acumulado do ano
df_fat_mes['faturamento_acumulado'] = (
    df_fat_mes['valor_final'].cumsum()
)

# ==========================================
# 4. EXIBIR A TABELA DE INDICADORES
# ==========================================

print('\nINDICADORES DE FATURAMENTO MENSAL\n')

print(
    df_fat_mes.round(2).to_string(index=False)
)

print('\nRESUMO ANUAL')
print(f"Faturamento total: R$ {df_fat_mes['valor_final'].sum():,.2f}")
print(f"Média mensal: R$ {media_mensal:,.2f}")

mes_maior = df_fat_mes.loc[
    df_fat_mes['valor_final'].idxmax()
]

mes_menor = df_fat_mes.loc[
    df_fat_mes['valor_final'].idxmin()
]

print(
    f"Maior faturamento: "
    f"{mes_maior['ano - mes'].strftime('%m/%Y')} "
    f"- R$ {mes_maior['valor_final']:,.2f}"
)

print(
    f"Menor faturamento: "
    f"{mes_menor['ano - mes'].strftime('%m/%Y')} "
    f"- R$ {mes_menor['valor_final']:,.2f}"
)

# ==========================================
# 5. GRÁFICO 1: FATURAMENTO MENSAL
# ==========================================

fig1 = px.line(
    df_fat_mes,
    x='ano - mes',
    y='valor_final',
    title='Evolução do Faturamento Mensal',
    markers=True,
    text='valor_final',
    labels={
        'ano - mes': 'Mês',
        'valor_final': 'Faturamento (R$)'
    }
)

fig1.update_traces(
    line=dict(width=3),
    marker=dict(size=8),
    texttemplate='R$ %{y:,.0f}',
    textposition='top center',
    hovertemplate=(
        '<b>%{x|%m/%Y}</b><br>'
        'Faturamento: R$ %{y:,.2f}'
        '<extra></extra>'
    )
)

fig1.add_hline(
    y=media_mensal,
    line_dash='dash',
    line_color='red',
    annotation_text=f'Média: R$ {media_mensal:,.2f}',
    annotation_position='bottom right'
)

fig1.update_layout(
    template='plotly_white',
    xaxis_title='Mês',
    yaxis_title='Faturamento (R$)',
    yaxis_tickprefix='R$ ',
    yaxis_tickformat=',.0f',
    hovermode='x unified'
)

fig1.show()

# ==========================================
# 6. GRÁFICO 2: VARIAÇÃO PERCENTUAL MENSAL
# ==========================================

df_variacao = df_fat_mes.dropna(
    subset=['variacao_%']
).copy()

df_variacao['desempenho'] = df_variacao[
    'variacao_%'
].apply(
    lambda x: 'Crescimento' if x >= 0 else 'Queda'
)

fig2 = px.bar(
    df_variacao,
    x='ano - mes',
    y='variacao_%',
    color='desempenho',
    text='variacao_%',
    title='Variação Percentual do Faturamento',
    color_discrete_map={
        'Crescimento': '#2CA25F',
        'Queda': '#D62728'
    },
    labels={
        'ano - mes': 'Mês',
        'variacao_%': 'Variação (%)',
        'desempenho': 'Resultado'
    }
)

fig2.update_traces(
    texttemplate='%{y:.1f}%',
    textposition='outside',
    cliponaxis=False,
    hovertemplate=(
        '<b>%{x|%m/%Y}</b><br>'
        'Variação: %{y:.2f}%'
        '<extra></extra>'
    )
)

fig2.add_hline(
    y=0,
    line_color='black',
    line_width=1
)

fig2.update_layout(
    template='plotly_white',
    xaxis_title='Mês',
    yaxis_title='Variação (%)',
    xaxis_tickformat='%m/%Y'
)

fig2.show()
```

```python

import pandas as pd
import plotly.express as px

# ==========================================
# 1. CRIAR DADOS FICTÍCIOS
# ==========================================

df_final = pd.DataFrame({
    'descricao_produto': [
        'Notebook', 'Notebook', 'Notebook',
        'Mouse', 'Mouse', 'Mouse',
        'Teclado', 'Teclado', 'Teclado',
        'Monitor', 'Monitor', 'Monitor',
        'Headset', 'Headset', 'Headset',
        'Webcam', 'Webcam', 'Webcam',
        'Impressora', 'Impressora', 'Impressora',
        'SSD', 'SSD', 'SSD',
        'Pen Drive', 'Pen Drive', 'Pen Drive'
    ],
    'valor_final': [
        3500, 4200, 3800,
        80, 100, 120,
        150, 180, 160,
        900, 1100, 950,
        180, 220, 200,
        120, 150, 130,
        650, 720, 680,
        350, 400, 380,
        40, 50, 60
    ],
    'lucro': [
        700, 900, 800,
        30, 40, 45,
        50, 65, 55,
        200, 280, 220,
        60, 80, 70,
        40, 55, 45,
        100, 130, 120,
        100, 130, 120,
        15, 20, 25
    ],
    'quantidade': [
        2, 2, 2,
        4, 5, 6,
        2, 3, 2,
        1, 1, 1,
        2, 2, 2,
        2, 3, 2,
        1, 1, 1,
        2, 2, 2,
        5, 6, 8
    ]
})

# ==========================================
# 2. AGRUPAR POR PRODUTO
# ==========================================

lucro_vf = (
    df_final.groupby('descricao_produto')
    .agg({
        'valor_final': 'sum',
        'lucro': 'sum',
        'quantidade': 'sum'
    })
    .reset_index()
)

# ==========================================
# 3. CALCULAR INDICADORES
# ==========================================

# Margem de lucro percentual
lucro_vf['margem_%'] = (
    lucro_vf['lucro'] / lucro_vf['valor_final']
) * 100

# Lucro médio por unidade vendida
lucro_vf['lucro_por_unidade'] = (
    lucro_vf['lucro'] / lucro_vf['quantidade']
)

# Participação de cada produto no faturamento
lucro_vf['participacao_%'] = (
    lucro_vf['valor_final']
    / lucro_vf['valor_final'].sum()
) * 100

# Ordenar pelo maior lucro
lucro_vf = lucro_vf.sort_values(
    by='lucro',
    ascending=False
)

# Médias para as linhas de referência
media_lucro = lucro_vf['lucro'].mean()
media_faturamento = lucro_vf['valor_final'].mean()

# ==========================================
# 4. EXIBIR RESUMO GERAL
# ==========================================

faturamento_total = lucro_vf['valor_final'].sum()
lucro_total = lucro_vf['lucro'].sum()
quantidade_total = lucro_vf['quantidade'].sum()

margem_geral = (
    lucro_total / faturamento_total
) * 100

print('=' * 45)
print('RESUMO GERAL DAS VENDAS')
print('=' * 45)

print(f'Faturamento total: R$ {faturamento_total:,.2f}')
print(f'Lucro total: R$ {lucro_total:,.2f}')
print(f'Quantidade vendida: {quantidade_total}')
print(f'Margem geral: {margem_geral:.2f}%')

print('\nDESEMPENHO POR PRODUTO')

print(
    lucro_vf[
        [
            'descricao_produto',
            'valor_final',
            'lucro',
            'quantidade',
            'margem_%',
            'participacao_%'
        ]
    ].round(2).to_string(index=False)
)

# ==========================================
# 5. GRÁFICO DE DISPERSÃO
# FATURAMENTO X LUCRO
# ==========================================

fig1 = px.scatter(
    lucro_vf,
    x='lucro',
    y='valor_final',
    color='descricao_produto',
    size='quantidade',
    size_max=35,
    hover_name='descricao_produto',
    hover_data={
        'lucro': ':.2f',
        'valor_final': ':.2f',
        'quantidade': True,
        'margem_%': ':.2f',
        'lucro_por_unidade': ':.2f',
        'participacao_%': ':.2f'
    },
    title='Desempenho dos Produtos: Faturamento x Lucro',
    labels={
        'descricao_produto': 'Produto',
        'lucro': 'Lucro total (R$)',
        'valor_final': 'Faturamento total (R$)',
        'quantidade': 'Unidades vendidas',
        'margem_%': 'Margem de lucro (%)',
        'lucro_por_unidade': 'Lucro por unidade (R$)',
        'participacao_%': 'Participação no faturamento (%)'
    }
)

# Linha vertical: lucro médio
fig1.add_vline(
    x=media_lucro,
    line_dash='dash',
    line_color='gray',
    annotation_text='Lucro médio',
    annotation_position='top'
)

# Linha horizontal: faturamento médio
fig1.add_hline(
    y=media_faturamento,
    line_dash='dash',
    line_color='gray',
    annotation_text='Faturamento médio',
    annotation_position='bottom right'
)

fig1.update_traces(
    marker=dict(
        opacity=0.8,
        line=dict(width=1, color='white')
    )
)

fig1.update_layout(
    template='plotly_white',
    title_x=0.5,
    xaxis_title='Lucro total (R$)',
    yaxis_title='Faturamento total (R$)',
    hovermode='closest',
    legend_title='Produtos'
)

fig1.show()

# ==========================================
# 6. GRÁFICO DE MARGEM DE LUCRO
# ==========================================

df_margem = lucro_vf.sort_values(
    by='margem_%',
    ascending=True
)

fig2 = px.bar(
    df_margem,
    x='margem_%',
    y='descricao_produto',
    orientation='h',
    color='margem_%',
    text='margem_%',
    color_continuous_scale='Viridis',
    title='Margem de Lucro por Produto',
    hover_data={
        'lucro': ':.2f',
        'valor_final': ':.2f',
        'quantidade': True
    },
    labels={
        'descricao_produto': 'Produto',
        'margem_%': 'Margem de lucro (%)',
        'lucro': 'Lucro total (R$)',
        'valor_final': 'Faturamento total (R$)',
        'quantidade': 'Unidades vendidas'
    }
)

fig2.update_traces(
    texttemplate='%{x:.1f}%',
    textposition='outside',
    cliponaxis=False
)

fig2.update_layout(
    template='plotly_white',
    title_x=0.5,
    xaxis_title='Margem de lucro (%)',
    yaxis_title='Produto',
    coloraxis_colorbar_title='Margem (%)'
)

fig2.show()

# ==========================================
# 7. GRÁFICO DE LUCRO TOTAL POR PRODUTO
# ==========================================

df_lucro = lucro_vf.sort_values(
    by='lucro',
    ascending=True
)

fig3 = px.bar(
    df_lucro,
    x='lucro',
    y='descricao_produto',
    orientation='h',
    color='lucro',
    text='lucro',
    color_continuous_scale='Blues',
    title='Ranking de Lucro por Produto',
    labels={
        'descricao_produto': 'Produto',
        'lucro': 'Lucro total (R$)'
    }
)

fig3.update_traces(
    texttemplate='R$ %{x:,.2f}',
    textposition='outside',
    cliponaxis=False
)

fig3.update_layout(
    template='plotly_white',
    title_x=0.5,
    xaxis_title='Lucro total (R$)',
    yaxis_title='Produto'
)

fig3.show()
```

```python

import pandas as pd
import numpy as np
import plotly.express as px

# ==================================================
# 1. CONFIGURAÇÕES E DADOS FICTÍCIOS
# ==================================================

np.random.seed(42)

produtos = [
    "Notebook", "Mouse", "Teclado", "Monitor", "Headset",
    "Webcam", "Impressora", "SSD", "HD Externo", "Pen Drive",
    "Tablet", "Smartphone", "Carregador", "Cabo USB", "Roteador",
    "Placa de Vídeo", "Processador", "Memória RAM", "Placa-Mãe",
    "Fonte ATX", "Gabinete", "Projetor", "Microfone",
    "Caixa de Som", "Smartwatch"
]

categorias = [
    "Computadores", "Periféricos", "Periféricos", "Monitores",
    "Áudio", "Periféricos", "Impressão", "Armazenamento",
    "Armazenamento", "Armazenamento", "Dispositivos móveis",
    "Dispositivos móveis", "Acessórios", "Acessórios", "Redes",
    "Componentes", "Componentes", "Componentes", "Componentes",
    "Componentes", "Componentes", "Imagem", "Áudio",
    "Áudio", "Dispositivos móveis"
]

# Preço de venda unitário fictício
precos = np.random.randint(20, 5001, size=len(produtos))

# Custo unitário entre 45% e 80% do preço de venda
custos = np.round(
    precos * np.random.uniform(0.45, 0.80, size=len(produtos)),
    2
)

# Dicionário com os dados dos produtos
cadastro_produtos = pd.DataFrame({
    "Produto": produtos,
    "Categoria": categorias,
    "Preco_Unitario": precos,
    "Custo_Unitario": custos
})

# ==================================================
# 2. GERAR VENDAS FICTÍCIAS
# ==================================================

registros = []

# Cada produto terá 8 vendas simuladas
for _, produto in cadastro_produtos.iterrows():

    for venda in range(8):

        quantidade = np.random.randint(1, 16)

        registros.append({
            "Produto": produto["Produto"],
            "Categoria": produto["Categoria"],
            "Preco_Unitario": produto["Preco_Unitario"],
            "Custo_Unitario": produto["Custo_Unitario"],
            "Quantidade": quantidade,
            "Venda": venda + 1
        })

df = pd.DataFrame(registros)

# ==================================================
# 3. CALCULAR FATURAMENTO, CUSTOS E LUCRO
# ==================================================

df["Faturamento"] = (
    df["Preco_Unitario"] * df["Quantidade"]
)

df["Custo_Total"] = (
    df["Custo_Unitario"] * df["Quantidade"]
)

df["Lucro"] = df["Faturamento"] - df["Custo_Total"]

# ==================================================
# 4. AGRUPAR OS DADOS POR PRODUTO
# ==================================================

resumo = df.groupby(
    ["Produto", "Categoria"],
    as_index=False
).agg(
    Quantidade_Vendida=("Quantidade", "sum"),
    Faturamento=("Faturamento", "sum"),
    Custo_Total=("Custo_Total", "sum"),
    Lucro=("Lucro", "sum")
)

# Margem percentual sobre o faturamento
resumo["Margem_Lucro_%"] = np.where(
    resumo["Faturamento"] > 0,
    resumo["Lucro"] / resumo["Faturamento"] * 100,
    0
)

# Participação percentual no faturamento geral
faturamento_total = resumo["Faturamento"].sum()

resumo["Participacao_Faturamento_%"] = (
    resumo["Faturamento"] / faturamento_total * 100
)

# Lucro médio por unidade vendida
resumo["Lucro_Por_Unidade"] = (
    resumo["Lucro"] / resumo["Quantidade_Vendida"]
)

# Ordenar do maior para o menor lucro
resumo = resumo.sort_values(
    by="Lucro",
    ascending=False
).reset_index(drop=True)

# ==================================================
# 5. APRESENTAR INDICADORES GERAIS
# ==================================================

quantidade_total = resumo["Quantidade_Vendida"].sum()
custo_total = resumo["Custo_Total"].sum()
lucro_total = resumo["Lucro"].sum()

margem_geral = (
    lucro_total / faturamento_total * 100
    if faturamento_total > 0 else 0
)

print("=" * 55)
print("       RELATÓRIO DE VENDAS — 25 PRODUTOS")
print("=" * 55)

print(f"Produtos analisados: {resumo['Produto'].nunique()}")
print(f"Registros de vendas: {len(df)}")
print(f"Unidades vendidas:   {quantidade_total}")
print(f"Faturamento total:   R$ {faturamento_total:,.2f}")
print(f"Custo total:         R$ {custo_total:,.2f}")
print(f"Lucro total:         R$ {lucro_total:,.2f}")
print(f"Margem geral:        {margem_geral:.2f}%")

print("\nTOP 10 PRODUTOS POR LUCRO:")
print(
    resumo[
        ["Produto", "Quantidade_Vendida", "Faturamento",
         "Lucro", "Margem_Lucro_%"]
    ].head(10).round(2).to_string(index=False)
)

# ==================================================
# 6. GRÁFICO 1 — FATURAMENTO X LUCRO
# ==================================================

fig1 = px.scatter(
    resumo,
    x="Faturamento",
    y="Lucro",
    color="Categoria",
    size="Quantidade_Vendida",
    hover_name="Produto",
    hover_data={
        "Quantidade_Vendida": True,
        "Margem_Lucro_%": ":.2f",
        "Faturamento": ":,.2f",
        "Lucro": ":,.2f"
    },
    title="Faturamento x Lucro por Produto",
    labels={
        "Faturamento": "Faturamento (R$)",
        "Lucro": "Lucro (R$)",
        "Categoria": "Categoria",
        "Quantidade_Vendida": "Unidades vendidas"
    }
)

fig1.update_layout(template="plotly_white")
fig1.show()

# ==================================================
# 7. GRÁFICO 2 — MARGEM DE LUCRO
# ==================================================

dados_margem = resumo.sort_values(
    "Margem_Lucro_%",
    ascending=True
)

fig2 = px.bar(
    dados_margem,
    x="Margem_Lucro_%",
    y="Produto",
    color="Categoria",
    orientation="h",
    text="Margem_Lucro_%",
    title="Margem de Lucro por Produto",
    labels={
        "Margem_Lucro_%": "Margem de lucro (%)",
        "Produto": "Produto"
    }
)

fig2.update_traces(
    texttemplate="%{x:.1f}%",
    textposition="outside",
    cliponaxis=False
)

fig2.update_layout(
    template="plotly_white",
    height=750
)

fig2.show()

# ==================================================
# 8. GRÁFICO 3 — RANKING DE LUCRO
# ==================================================

dados_lucro = resumo.sort_values(
    "Lucro",
    ascending=True
)

fig3 = px.bar(
    dados_lucro,
    x="Lucro",
    y="Produto",
    color="Categoria",
    orientation="h",
    title="Ranking de Lucro por Produto",
    labels={
        "Lucro": "Lucro total (R$)",
        "Produto": "Produto"
    },
    hover_data={
        "Faturamento": ":,.2f",
        "Custo_Total": ":,.2f",
        "Margem_Lucro_%": ":.2f"
    }
)

fig3.update_layout(
    template="plotly_white",
    height=750
)

fig3.show()

# ==================================================
# 9. EXPORTAR PARA EXCEL
# ==================================================

with pd.ExcelWriter(
    "analise_25_produtos.xlsx",
    engine="openpyxl"
) as arquivo:

    df.to_excel(
        arquivo,
        sheet_name="Vendas_Detalhadas",
        index=False
    )

    resumo.to_excel(
        arquivo,
        sheet_name="Resumo_Produtos",
        index=False
    )

    cadastro_produtos.to_excel(
        arquivo,
        sheet_name="Cadastro",
        index=False
    )

print("\nArquivo Excel gerado: analise_25_produtos.xlsx")
```
