```python
#juncao do dados
#concatenar

df_vendas = pd.concat([df_vendas1,df_vendas2], axis=0, join='outer' )#
df_vendas.head()

```

```python
#mesclagem

df_vendas['mes'] = df_vendas['dt_venda'].dt.month
df_vendas['ano'] = df_vendas['dt_venda'].dt.year
```

```python
df_vendas.info()
```

```python
df_final = df_vendas.merge(df_metas, on=['ano', 'mes'],how='inner')
```

```python
df_final.info()
```

```python
df_final['lucro'] = (df_final['valor_unitario'] - df_final['preco_custo']) * df_final['quantidade']
df_final['margem_lucro'] = (df_final['valor_unitario'] - df_final['preco_custo']) / df_final['valor_unitario']
df_final['prazo_entrega'] = df_final['dt_entrega'] - df_final['dt_venda']
df_final['valor_final'] = df_final['valor_unitario'] * df_final['quantidade']
```

```python
df_final.drop(columns=['matricula_funcionario','codigo_produto','codigo_loja'],inplace=True)
```

```python
#AGRUPAMENTO - reset_index serve para deixar extruturado o titulo e o sort.values serve para deixa do maior para o menor o true deixa menor e false maior

vendedores_vf = df_final.groupby('nome_funcionario')['valor_final'].sum().sort_values(ascending=False).reset_index()

display(vendedores_vf)

```

```python
#agrupar por vendedor e somar tambem

vendedores_vf_qtd = df_final.groupby('nome_funcionario')[['valor_final','lucro']].sum().sort_values(by='lucro',ascending=False).reset_index()

display(vendedores_vf_qtd)
```

```python
#quem mais vendeu por cada produto

vendedores_vf_qtd = df_final.groupby(['nome_funcionario','descricao_produto'])['quantidade'].sum()

display(vendedores_vf_qtd)
```

```python
#quem mais vendeu o produto sozinho

vendedores_vf_qtd = df_final.groupby(['nome_funcionario','descricao_produto'])['quantidade'].sum().sort_values(ascending=False).reset_index()

display(vendedores_vf_qtd)
```

```python
vendas_vf_qtd = df_final.groupby('nome_funcionario').agg({
    'valor_final':['sum','mean'],
    'quantidade':['min','max','mean']
}).reset_index()

display(vendas_vf_qtd)
```
