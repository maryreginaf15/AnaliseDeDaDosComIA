```python
import pandas as pd
```

```python
#importacao de dados


#df_vendas1 = pd.read_csv("/content/base_dados_vendas_tratada.csv")
#df_vendas2 = pd.read_csv("/content/dados_complementares.csv")
#df_metas = pd.read_excel("/content/Python_JOIN_Metas.xlsx")
df_vendas1 = pd.read_csv("/content/base_dados_vendas_tratada.csv",parse_dates = ['dt_venda', 'dt_entrega'])
df_vendas2 = pd.read_csv("/content/dados_complementares.csv",parse_dates = ['dt_venda', 'dt_entrega'])
df_metas = pd.read_excel("/content/Python_JOIN_Metas.xlsx")
```

```python
#juncao do dados
#concatenar

df_vendas = pd.concat([df_vendas1,df_vendas2], axis=0, join='outer' )#
df_vendas.head()

```

```python
df_vendas.info()
```

```python
display(df_vendas)
```

```python
df_vendas.duplicated().sum()
```

```python
#mesclagem

df_vendas['mes'] = df_vendas['dt_venda'].dt.month
df_vendas['ano'] = df_vendas['dt_venda'].dt.year
```

```python
df_vendas.info()
```

```python
df_final = df_vendas.merge(df_metas, on=['ano', 'mes'],how='inner')
```

```python
df_final.info()
```

```python
df_final['lucro'] = (df_final['valor_unitario'] - df_final['preco_custo']) * df_final['quantidade']
df_final['margem_lucro'] = (df_final['valor_unitario'] - df_final['preco_custo']) / df_final['valor_unitario']
df_final['prazo_entrega'] = df_final['dt_entrega'] - df_final['dt_venda']
df_final['valor_final'] = df_final['valor_unitario'] * df_final['quantidade']
```

```python
df_final.drop(columns=['matricula_funcionario','codigo_produto','codigo_loja'],inplace=True)
```

```python
df_final.info()
```

```python
#salvando a base

df_final.to_csv("/content/base_dados_tratada.csv")
```
