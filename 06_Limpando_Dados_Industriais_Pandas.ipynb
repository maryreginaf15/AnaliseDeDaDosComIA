## 1. Series e DataFrame

### 📘 Conceito teórico

O Pandas tem duas estruturas principais:

- **Series**: parecida com uma lista ou um array, mas com um **índice nomeado** para cada valor — como uma única coluna de uma planilha.
- **DataFrame**: uma **tabela completa**, com várias colunas (cada uma sendo uma Series) — como uma planilha inteira.

```
DataFrame:
┌─────────────────┬───────────┬────────────┐
│ maquina          │ setor     │ producao   │
├─────────────────┼───────────┼────────────┤
│ Torno CNC 03     │ Usinagem  │ 480        │  ← linha 0
│ Prensa 12        │ Estamparia│ 320        │  ← linha 1
└─────────────────┴───────────┴────────────┘
        ↑ cada coluna é uma Series
```

### 🔎 Exemplo extremamente simples


````python
import pandas as pd

producao = pd.Series([480, 320, 610])
print(producao)
````

### 🔍 Explicação linha por linha

| Código | O que significa |
|---|---|
| `import pandas as pd` | Importa o Pandas, apelidado de `pd` (convenção usada por quase todo mundo). |
| `pd.Series([...])` | Cria uma Series a partir de uma lista. |

**Perceba que** ao imprimir a Series, aparece uma coluna de números à esquerda (`0, 1, 2`) — esse é o **índice**, criado automaticamente pelo Pandas.

### 🏭 Exemplo aplicado à indústria: criando um DataFrame pequeno


````python
dados = {
    "nome": ["Torno CNC 03", "Prensa 12", "Solda Robô 01"],
    "setor": ["Usinagem", "Estamparia", "Soldagem"],
    "producao": [480, 320, 610]
}

df_maquinas = pd.DataFrame(dados)
print(df_maquinas)
````

**Observe que** criamos primeiro um **dicionário**, onde cada chave vira o nome de uma coluna, e cada lista de valores vira o conteúdo daquela coluna. O `pd.DataFrame()` transforma esse dicionário numa tabela organizada. Isso funciona bem para tabelas pequenas — mas, para uma base de centenas de linhas como a que usaremos no restante desta missão, o caminho é ler diretamente de um arquivo CSV, como veremos a seguir.

### 📦 O que você aprendeu até aqui?

| Conceito | Em palavras simples |
|---|---|
| Series | Uma única coluna de dados, com índice |
| DataFrame | Uma tabela completa, com várias colunas |

## 3. Explorando um DataFrame

### 📘 Conceito teórico

Antes de qualquer análise, é comum "dar uma olhada geral" no DataFrame: quantas linhas e colunas tem, quais os tipos de cada coluna, quais os primeiros registros. Com uma base de 543 linhas, essa etapa se torna ainda mais importante — seria impossível conferir tudo "no olho".

### 📦 O que você aprendeu até aqui?

## 4. Selecionando Colunas e Linhas

### 📘 Conceito teórico

Assim como acessamos itens de uma lista pelo índice, no Pandas selecionamos **colunas pelo nome** e **linhas por posição ou por condição**.

### 🔎 Exemplo extremamente simples: selecionando uma coluna


````python
print(df["maquina"])
````

**Observe que** selecionar uma única coluna com `df["maquina"]` retorna uma **Series** com as 543 linhas — lembre-se, uma coluna isolada de um DataFrame é justamente uma Series.

### 🏭 Exemplo aplicado à indústria: selecionando linhas por posição


````python
print(df.iloc[0])      # primeira linha (posição 0)
print(df.iloc[0:3])    # as três primeiras linhas
````

| Código | O que significa |
|---|---|
| `.iloc[posicao]` | Seleciona linha(s) pela **posição numérica**, começando em 0 — o `i` de `iloc` remete a "index location". |

### 📦 O que você aprendeu até aqui?

| Código | O que faz |
|---|---|
| `df["coluna"]` | Seleciona uma coluna específica |
| `df.iloc[n]` | Seleciona a linha na posição `n` |

## 5. Filtros Básicos

### 📘 Conceito teórico

Um **filtro** seleciona apenas as linhas de um DataFrame que atendem a uma determinada condição — de forma parecida com o `if` que já conhecemos, mas aplicado a uma tabela inteira de uma vez.

### 🏠 Analogia do cotidiano

É como pedir para um assistente: "me mostre só os dias em que a produção foi maior que 480 peças" — em vez de você mesmo olhar, uma por uma, as 543 linhas da tabela.

### 🔎 Exemplo extremamente simples


````python
dias_alta_producao = df[df["producao"] > 480]
print(dias_alta_producao.shape)
dias_alta_producao.head()
````

### 🔍 Explicação detalhada do resultado

| Código | O que significa |
|---|---|
| `df["producao"] > 480` | Cria uma Series de `True`/`False`, uma para cada uma das 543 linhas, indicando se a condição é satisfeita. |
| `df[ ... ]` | Ao colocar essa Series de True/False dentro dos colchetes do DataFrame, o Pandas retorna **apenas as linhas onde o valor é `True`** — neste caso, 224 das 543 linhas. |

**É comum iniciantes confundirem** este uso dos colchetes com a seleção de uma coluna (`df["producao"]`). A diferença está no que vai dentro: um **nome de coluna** (texto) seleciona uma coluna; uma **condição** (que gera True/False) filtra linhas.

### 🏭 Exemplo aplicado à indústria: combinando filtro e cálculo


````python
df["taxa_defeito"] = (df["pecas_defeituosas"] / df["producao"]) * 100
df[["maquina", "producao", "pecas_defeituosas", "taxa_defeito"]].head()
````

**Perceba que** criamos uma **nova coluna** (`taxa_defeito`) simplesmente atribuindo um cálculo a `df["taxa_defeito"]` — o Pandas calcula essa nova coluna para as 543 linhas de uma vez, sem precisar de um `for`.

### 📦 O que você aprendeu até aqui?

| Conceito | Em palavras simples |
|---|---|
| Filtro | Seleciona apenas as linhas que atendem a uma condição |
| Nova coluna calculada | Criar uma coluna a partir de operações com outras colunas existentes |

## 🚧 Erros que iniciantes costumam cometer

1. **Confundir seleção de coluna com filtro de linhas.**
   `df["producao"]` seleciona uma coluna; `df[df["producao"] > 480]` filtra linhas. A diferença está em usar um nome de coluna ou uma condição dentro dos colchetes.

2. **Esquecer aspas ao selecionar uma coluna.**
   Escrever `df[producao]` sem aspas faz o Python procurar uma **variável** chamada `producao`, não a coluna. O correto é `df["producao"]`.

3. **Confundir `.iloc[]` (posição) com seleção por nome de índice.**
   `.iloc[0]` sempre pega a primeira linha pela posição, mesmo que os índices do DataFrame tenham sido reorganizados.

4. **Esquecer que `df.head()` só mostra uma amostra.**
   Com 543 linhas, ver que os primeiros registros "parecem certos" não garante que o resto da tabela também esteja correto — vale sempre checar `.info()` e `.describe()` também.

5. **Tentar imprimir o DataFrame inteiro sem limitar a visualização.**
   Digitar apenas `df` numa tabela de 543 linhas gera uma saída enorme e pouco útil — prefira `df.head()` para uma primeira olhada.
