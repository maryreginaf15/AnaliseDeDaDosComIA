## 1. O laço `for`

### 📘 Conceito teórico

Um **laço de repetição** (ou *loop*) permite executar o mesmo bloco de código várias vezes, sem precisar copiar e colar. O `for` repete um bloco **um número definido de vezes**, geralmente percorrendo uma sequência de valores.

### 🏠 Analogia do cotidiano

Pense numa esteira de inspeção: o inspetor repete o mesmo procedimento — "pegar peça, verificar, colocar de volta" — para cada peça que passa. Ele não escreve um procedimento novo para cada peça; repete o mesmo processo várias vezes.

### 🔎 Exemplo extremamente simples


````python
for numero in range(5):
    print(numero)
````

### 🔍 Explicação linha por linha

| Código | O que significa |
|---|---|
| `range(5)` | É uma **função** que gera uma sequência de números: `0, 1, 2, 3, 4` (começa em 0 e para **antes** de chegar em 5). |
| `for numero in range(5):` | Para cada valor dentro de `range(5)`, guarde-o temporariamente na variável `numero` e execute o bloco abaixo. |
| `    print(numero)` | Executado uma vez para cada valor — por isso vemos 5 números impressos. |

**É comum iniciantes confundirem** o `range(5)` achando que ele vai de 1 a 5. Na verdade, ele começa em 0 e tem exatamente 5 números: 0, 1, 2, 3 e 4.

### 🏭 Exemplo aplicado à indústria


````python
for leitura in range(5):
    print("Leitura do sensor número", leitura + 1)
````

**Perceba que** somamos `+ 1` só para exibir "Leitura 1, 2, 3..." em vez de começar do zero — isso não muda o funcionamento do `range`, só a forma como mostramos o resultado.

### 📦 O que você aprendeu até aqui?

| Conceito | Em palavras simples |
|---|---|
| `for` | Repete um bloco de código um número definido de vezes |
| `range(n)` | Gera uma sequência de números de 0 até n-1 |

## 2. O laço `while`

### 📘 Conceito teórico

O `while` repete um bloco de código **enquanto uma condição for verdadeira** — diferente do `for`, que repete um número já definido de vezes.

### 🏠 Analogia do cotidiano

É como encher um balde com água: você continua enchendo **enquanto** o balde não estiver cheio. Você não sabe de antemão exatamente quantos segundos vai levar — só sabe que vai parar quando a condição "balde cheio" for atingida.

### 🔎 Exemplo extremamente simples


````python
contador = 0

while contador < 5:
    print(contador)
    contador = contador + 1
````

### 🔍 Explicação linha por linha

| Código | O que significa |
|---|---|
| `contador = 0` | Cria a variável que vamos usar para controlar quando parar. |
| `while contador < 5:` | Enquanto essa condição for `True`, o bloco abaixo se repete. |
| `contador = contador + 1` | Aumenta o valor de `contador` em 1 a cada repetição — **essencial** para que a condição um dia se torne falsa. |

**É comum iniciantes esquecerem** de atualizar a variável de controle (`contador = contador + 1`). Se isso for esquecido, a condição nunca se torna falsa, e o programa fica preso num **loop infinito**, repetindo para sempre. Se isso acontecer no Jupyter, você pode interromper a execução clicando no botão de "parar" (quadrado) na barra de ferramentas.

### 🏭 Exemplo aplicado à indústria


````python
nivel_tanque = 0

while nivel_tanque < 100:
    nivel_tanque = nivel_tanque + 20
    print("Nível do tanque:", nivel_tanque, "%")

print("Tanque cheio!")
````

### 📦 O que você aprendeu até aqui?

| Conceito | Em palavras simples |
|---|---|
| `while` | Repete enquanto uma condição for verdadeira |
| Risco do `while` | Esquecer de atualizar a condição gera um loop infinito |

## 3. Listas

### 📘 Conceito teórico

Uma **lista** é uma coleção **ordenada** de valores, guardados numa única variável. Cada valor tem uma posição, chamada de **índice**, que começa em `0`.

### 🏠 Analogia do cotidiano

Pense numa prateleira numerada de peças: a posição 0 é a primeira prateleira, a posição 1 é a segunda, e assim por diante. Cada prateleira guarda uma peça diferente, mas todas fazem parte do mesmo armário (a lista).

```
Lista:      [ 23.5,  24.1,  22.8,  25.0 ]
Índice:        0      1      2      3
```

### 🔎 Exemplo extremamente simples


````python
temperaturas = [23.5, 24.1, 22.8, 25.0]

print(temperaturas)
print(temperaturas[0])
print(temperaturas[2])
````

### 🔍 Explicação linha por linha

| Código | O que significa |
|---|---|
| `[23.5, 24.1, 22.8, 25.0]` | Os colchetes `[ ]` criam uma lista; cada valor é separado por vírgula. |
| `temperaturas[0]` | Acessa o valor na posição de **índice 0** — a primeira posição da lista, que é `23.5`. |
| `temperaturas[2]` | Acessa o valor na posição de índice 2 — a **terceira** posição (`22.8`), já que a contagem começa em 0. |

**É comum iniciantes confundirem** índice com "posição na ordem do dia a dia". Na programação, contamos a partir de 0 — então o "terceiro" item está no índice 2, não no índice 3.

### 🏭 Exemplo aplicado à indústria


````python
leituras_sensor = [78.2, 79.0, 77.5, 80.1, 78.9]

leituras_sensor.append(81.3)
print("Leituras após novo registro:", leituras_sensor)

print("Total de leituras:", len(leituras_sensor))
print("Última leitura:", leituras_sensor[-1])
````

**Perceba que**:
- `.append(81.3)` é um **método** — uma função que "pertence" à lista — usado para adicionar um novo valor ao final.
- `len(leituras_sensor)` retorna quantos itens existem na lista.
- `leituras_sensor[-1]` usa um **índice negativo**, que conta a partir do final da lista: `-1` é sempre o último item, `-2` o penúltimo, e assim por diante.

### 📦 O que você aprendeu até aqui?

| Conceito | Em palavras simples |
|---|---|
| Lista | Coleção ordenada de valores, guardada em uma variável |
| Índice | Posição de um item na lista, começando em 0 |
| `.append()` | Adiciona um item ao final da lista |
| `len()` | Conta quantos itens existem na lista |

## 4. Tuplas

### 📘 Conceito teórico

Uma **tupla** é parecida com uma lista — também guarda vários valores em ordem — mas, uma vez criada, **não pode ser alterada** (não é possível adicionar, remover ou modificar itens).

### 🏠 Analogia do cotidiano

Pense numa tupla como a placa de identificação fixada numa máquina: modelo, número de série, ano de fabricação. Essas informações não mudam depois que a placa foi fixada — diferente de uma lista de peças produzidas, que muda o tempo todo.

### 🔎 Exemplo extremamente simples


````python
coordenadas = (10, 20)
print(coordenadas)
print(coordenadas[0])
````

**Observe que** tuplas usam parênteses `( )` em vez dos colchetes `[ ]` das listas. O acesso por índice funciona igual ao das listas.

### 🏭 Exemplo aplicado à indústria


````python
info_maquina = ("Torno CNC 03", 2019, "Setor A")

print("Nome:", info_maquina[0])
print("Ano de fabricação:", info_maquina[1])
print("Setor:", info_maquina[2])
````

**É comum iniciantes perguntarem** "por que usar tupla se a lista já faz tudo isso?". A resposta é: quando queremos **garantir** que aquela informação não seja alterada por engano em algum ponto do código, a tupla é mais segura, porque ela simplesmente não permite alteração.

### 📦 O que você aprendeu até aqui?

| Conceito | Em palavras simples |
|---|---|
| Tupla | Como uma lista, mas que não pode ser alterada depois de criada |
| Quando usar | Para dados fixos, que não devem mudar (ex: identificação de uma máquina) |

## 5. Dicionários

### 📘 Conceito teórico

Um **dicionário** guarda valores associados a uma **chave** (um nome descritivo), em vez de um índice numérico. É como uma lista, mas em que cada item tem um "rótulo" em vez de uma posição.

### 🏠 Analogia do cotidiano

É como uma ficha de cadastro de uma máquina: em vez de "posição 0, posição 1, posição 2", temos campos nomeados: "nome", "setor", "ano". Você busca a informação pelo nome do campo, não pela posição.

```
Dicionário:
{
  "nome":  "Torno CNC 03",
  "setor": "Usinagem",
  "ano":   2019
}
```

### 🔎 Exemplo extremamente simples


````python
maquina = {"nome": "Torno CNC 03", "setor": "Usinagem", "ano": 2019}

print(maquina["nome"])
print(maquina["ano"])
````

### 🔍 Explicação linha por linha

| Código | O que significa |
|---|---|
| `{ }` | As chaves `{ }` criam um dicionário. |
| `"nome": "Torno CNC 03"` | Um **par chave-valor**: `"nome"` é a chave, `"Torno CNC 03"` é o valor associado a ela. |
| `maquina["nome"]` | Acessa o valor associado à chave `"nome"`, usando colchetes (mas com o nome da chave, não um número de índice). |

**É comum iniciantes confundirem** dicionário com lista por ambos usarem colchetes para acessar valores. A diferença é: na lista, o que vai dentro dos colchetes é um número (índice); no dicionário, é uma chave (geralmente um texto).

### 🏭 Exemplo aplicado à indústria


````python
sensor = {"id": "S-102", "localizacao": "Forno 2", "temperatura": 850, "ativo": True}

print("Sensor", sensor["id"], "está no local:", sensor["localizacao"])
print("Temperatura atual:", sensor["temperatura"])

sensor["temperatura"] = 862
print("Temperatura atualizada:", sensor["temperatura"])
````

**Perceba que**, diferente das tuplas, os valores de um dicionário **podem ser alterados** depois de criados — só a estrutura de "chave associada a um valor" se mantém.

### 📦 O que você aprendeu até aqui?

| Conceito | Em palavras simples |
|---|---|
| Dicionário | Guarda valores associados a chaves nomeadas, em vez de índices numéricos |
| Par chave-valor | A unidade básica de um dicionário: `"chave": valor` |

## 6. Conjuntos

### 📘 Conceito teórico

Um **conjunto** (`set`) guarda valores **sem repetição** e **sem ordem definida**. É útil quando você quer garantir que não existam itens duplicados.

### 🏠 Analogia do cotidiano

Pense num conjunto como uma lista de códigos de defeitos únicos encontrados numa inspeção: mesmo que o mesmo tipo de defeito apareça em 10 peças diferentes, você só quer saber **quais tipos diferentes** de defeito ocorreram, sem repetir o mesmo tipo várias vezes.

### 🔎 Exemplo extremamente simples


````python
numeros = {1, 2, 2, 3, 3, 3}
print(numeros)
````

**Observe que**, mesmo colocando `2` duas vezes e `3` três vezes, o conjunto guarda apenas um exemplar de cada valor. Conjuntos também usam chaves `{ }`, mas sem os dois-pontos de chave-valor dos dicionários.

### 🏭 Exemplo aplicado à indústria


````python
defeitos_encontrados = {"risco", "amassado", "risco", "trinca", "amassado"}
print("Tipos de defeito distintos encontrados:", defeitos_encontrados)
print("Quantidade de tipos diferentes:", len(defeitos_encontrados))
````

### 📦 O que você aprendeu até aqui?

| Conceito | Em palavras simples |
|---|---|
| Conjunto (`set`) | Guarda valores sem repetição, sem ordem garantida |
| Quando usar | Quando você só se importa com valores únicos, não com duplicatas |

## 🚧 Erros que iniciantes costumam cometer

1. **Esquecer de atualizar a variável de controle no `while`.**
   Isso gera um loop infinito, pois a condição nunca se torna falsa.

2. **Confundir índice com "posição do dia a dia".**
   O primeiro item de uma lista está no índice `0`, não no índice `1`.

3. **Tentar alterar uma tupla.**
   Escrever `coordenadas[0] = 15` gera erro, porque tuplas não podem ser modificadas após criadas.

4. **Usar índice numérico num dicionário.**
   Escrever `maquina[0]` ao invés de `maquina["nome"]` gera erro, a menos que `0` seja de fato uma das chaves.

5. **Esperar que um conjunto mantenha a ordem de inserção.**
   Ao imprimir um `set`, a ordem dos elementos pode não ser a mesma em que foram digitados — isso é esperado e normal.

## ✅ Resumo final da Missão 3

- Dicionários guardam valores associados a chaves nomeadas.
- Conjuntos guardam valores únicos, sem repetição.
