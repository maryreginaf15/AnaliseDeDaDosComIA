```python
#tratamento de  horas e datas

df_vendas['dt_venda'] = pd.to_datetime(df_vendas['dt_venda'], format='%d/%m/%Y %H:%M')
df_vendas['dt_entrega'] = pd.to_datetime(df_vendas['dt_entrega'], format='%d/%m/%Y %H:%M')
```

```python
df_vendas['preco_custo']= df_vendas['preco_custo'].str.replace(',','.')
df_vendas['valor_unitario']= df_vendas['valor_unitario'].str.replace(',','.')
df_vendas['comissao']= df_vendas['comissao'].str.replace(',','.')

df_vendas
```

```python
df_vendas['preco_custo']= df_vendas['preco_custo'].astype(float)
df_vendas['valor_unitario']= df_vendas['valor_unitario'].astype(float)
df_vendas['comissao']= df_vendas['comissao'].astype(float)
df_vendas
```

```python
df_vendas['comissao']
```

```python
#preencher os valores nulos com a media

valor_medio = df_vendas['comissao'].mean()
print(valor_medio)

df_vendas['comissao'].fillna(valor_medio, inplace=True)
```

```python
valor_moda = df_vendas['comissao'].mode()
print(valor_moda)

df_vendas['comissao'].fillna(valor_moda, inplace=True)
```

```python
#preenchendo so valores de baixo para cimar


df_vendas['comissao'].ffill(inplace=True)
#df_vendas['comissao'].fillna(method='bfill', inplace=True)
```

```python
#preenchendo os valores com bfiff
df_vendas['comissao'].bfill(inplace=True)
```

```python
#remover matriculas de funcionarios e codigo do produto

df_vendas.drop(columns=['matricula_funcionario', 'codigo_produto'],inplace=True)
```

```python
df_vendas.info()
```

```python
#engenharia de dados

df_vendas['valor_final'] = df_vendas['quantidade' ]* df_vendas['valor_unitario']
df_vendas['valor_comissao'] = df_vendas ['valor_final' ]* df_vendas['comissao']
df_vendas['lucro_bruto'] = (df_vendas ['valor_final' ]* df_vendas['preco_custo']) * df_vendas['quantidade']
df_vendas['lucro_real'] = df_vendas ['lucro_bruto' ] -  df_vendas['valor_comissao']
df_vendas['margem_lucro'] = (df_vendas ['valor_unitario' ] -  df_vendas['preco_custo'] ) / df_vendas['preco_custo']/ df_vendas['valor_unitario'] * 100
df_vendas['markup'] =  (df_vendas['valor_unitario'] - df_vendas['preco_custo']) / df_vendas['preco_custo'] * 100
```
